```python
# FILTERING
```

```python
# Select elements from an array that match a given condition 

ages = np.array([[21,34,16,22,47,99,27],[55,50,19,16,18,26,22]])
teens = ages[ages < 18]
print(teens)
```

```python
adults = ages[(18<ages) & (ages<40)]
print(adults)
```

```python
seniors = ages[ages > 40]
print(seniors)
```

```python
odd_ages = ages[ages % 2 != 0]
print(odd_ages)
```

```python
# Where function : returns a same shape array with filling (0/-1) where the condition is not satisfied
adults = np.where(ages >= 18 , ages , -1) 
print(adults)
```

```python
# RANDOM NUMBERS
```

```python
# Random number generator (default_rng)

rng = np.random.default_rng()

print(rng.integers(low = 1, high = 101 , size = (3,2))) 
# rng.integers : returns random integer in a range (second arg is exclusive) , size = amount of numbers
print(np.random.uniform(low = -1 , high = 5 , size = 3)) 
# np.random.uniform : random floats
```

```python
# Suffling an array 
array = np.array([1,2,3,4,5,6,7])
print("Array before shuffling : " , array)

rng.shuffle(array)
print("Array after shuffling : " , array)
```

```python
# Choosing a random element of an array 

cats = np.array(["lion" , "tiger" , "snow leopard" ,"saber cat" , "domestic cat" , "Black footed cat" ])

print("YOUR CAT IS : " , rng.choice(cats))
print("YOUR CATs ARE : " , rng.choice(cats , size = 3))
```

```python
# For the same random result we can set a seed to the generator 

seed_rng = np.random.default_rng(seed = 1)

print("Your variable random cat : " , rng.choice(cats))
print("Your permanent random cat : " , seed_rng.choice(cats))
```
