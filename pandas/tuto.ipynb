```python
# Filtering 

df[(df["Country"] == 'USA' ) & (df["Category"] != "Electronics") ]
```

```python
# Advanced filtering 

df[df["CustomerName"].str.startswith("Z")]
```

```python
df[df["Country"].isin(["Sweeden" , "France" , "Germany"])]
```

```python
# tilda operator : ~ 
# if we add the operator in the begining of the filter , it's negation will be considered (the inverse)
df[~df["Country"].isin(["Sweeden" , "France" , "Germany"])]
```

```python
#Updating Data 
```

```python
# locating based on a label 
# a label being any antry in a row 

df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
# Changing a value after loc 

df.loc[df["CustomerName"] == "Yuki Yamamoto" , "Quantity"] = 12
df.loc[df["CustomerName"] == "Yuki Yamamoto"]
```

```python
# Changing values of an entire column 
 
df["Country"] =  df["Country"].str.upper()
```

```python
print(set(df["Country"]))
```

```python
# DELETING DATA

#deleting by index
df = df.drop(39)
```

```python
df.tail()
```

```python
# Dropping null values
df.dropna()
```

```python
# Filling null values 
df.fillna({"OrderID" : -1} , inplace = True)
# argument "inplace : True" means it's going to modify thexisting dataframe 
# and not create a new one 
```

```python
#renaming columns 
df.rename(columns ={"OrderID" : "Order ID"} , inplace = True)
```

```python
df["Country"].value_counts()
#counts the occurences of a value of a specific column
```

```python
# Groupong by certain attributes 
df.groupby("Country")
```

```python
(df.groupby(["Country" , "Product"])["Price"].sum() * df.groupby(["Country" , "Product"])["Quantity"].sum()) 
# Searching the values (money) generated from each product in each country 
```

```python
(df.groupby(["Category","Country"])["Price"].sum() * df.groupby(["Category","Country"])["Quantity"].sum()).sort_values(ascending=False)
# sort_values() , sorting values in ascending or descending order 

# Values generated by each category in each country 
```
