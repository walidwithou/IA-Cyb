```python
# CATEGORIZATION 
```

```python
# pandas.qcut() : used to discretize (segment , section) continuous numeric data into 
# categorical bins based on sample quantiles (percentilles) 
# qcut devides data such as each bin contains the same number of observations

df["PriceCat"] = pd.qcut(df["Price"] , 3 , labels= ["Cheap" , "Medium" , "Expensive"])

print(df[df["PriceCat"] == "Expensive"].value_counts().sum() , df[df["PriceCat"] == "Medium"].value_counts().sum() , df[df["PriceCat"] == "Cheap"].value_counts().sum())
```

```python
# Row / column wise functions 
# functions that apply on the dataframe row by row or column by column 

# Row wise : axis = 1 

df['TotalPrice'] = df.apply(lambda row : row['Price'] * row['Quantity'] , axis=1)

print(df['TotalPrice'].tail())
```

```python
# Column wise 
df.apply(lambda col : col.max() , axis = 0)
```
